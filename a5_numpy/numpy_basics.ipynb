```python
import matplotlib.pyplot as plt 

fig, ax = plt.subplots(1)

x = np.linspace(-10,10)
y = x**2

ax.plot(x,y)
ax.set(title="f(x)=$x^2$", xlabel="x", ylabel="y")
```

## array slicing

```python
v = np.array([1, 2, 34, 1, 6, 3])
print(f"{v = }") 
print(f"{v[2] = }") 
print(f"{v[0] = }") 
print(f"{v[2:] = }") 
print(f"{v[2:4] = }") 
```
